```python
import pandas as pd
from transformers import T5Tokenizer, Trainer, TrainingArguments, T5ForConditionalGeneration
```

```python
train_data = pd.read_csv('/content/drive/MyDrive/ColabNotebook/samsum-train.csv')
val_data = pd.read_csv('/content/drive/MyDrive/ColabNotebook/samsum-validation.csv')
```

# random sampling

```python
train_data = train_data.sample(n= 4000, random_state= 42).reset_index(drop= True)
val_data = val_data.sample(n= 500, random_state= 42).reset_index(drop= True)
```

# data cleaning

```python
import re

def clean_data(text):
    text = re.sub(r'\r\n', ' ', text)  # lines
    text = re.sub(r'\s+', ' ', text)  # spaces
    text = re.sub(r'<.*?>', ' ', text)  # html tags
    text = text.strip().lower()
    return text
```

```python
train_data['dialogue'] = train_data['dialogue'].apply(clean_data)
train_data['summary'] = train_data['summary'].apply(clean_data)

val_data['dialogue'] = val_data['dialogue'].apply(clean_data)
val_data['summary'] = val_data['summary'].apply(clean_data)
```

# tokenization for T5

```python
tokenizer = T5Tokenizer.from_pretrained('t5-small')
```

```python
# raw data => tokenized inputs for fine-tuning
def tokenize(data):
    inputs = tokenizer(data['dialogue'], padding= "max_length", max_length= 512, truncation= True)
    targets = tokenizer(data['summary'], padding= "max_length", max_length= 150, truncation= True)
    inputs['labels'] = targets['input_ids'] # token is added to inputs as labels
    return inputs
```

```python
train_dataset = train_data.apply(tokenize, axis= 1).tolist()
val_dataset = val_data.apply(tokenize, axis= 1).tolist()
```

```python
train_dataset[0]
```

# fine tuning of our transformer model

```python
# nlp => generation task
model = T5ForConditionalGeneration.from_pretrained('t5-small')
```

```python
import torch
```

```python
if torch.backends.mps.is_available():
    device= torch.device('mps')
elif torch.cuda.is_available():
    device= torch.device('cuda')
else:
    device = torch.device('cpu')

print("Device = ", device)
model.to(device)
```

```python
# define training arguments
training_args = TrainingArguments(
    output_dir= './drive/MyDrive/ColabNotebook/results',
    num_train_epochs= 6,
    weight_decay= 0.01,
    per_device_train_batch_size= 8,
    per_device_eval_batch_size= 8,
    eval_strategy= "epoch",
    save_strategy= "epoch",
    warmup_steps= 500    #  0 => lr default
)
```

```python
trainer= Trainer(
    model = model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=val_dataset
)
```

```python
# train the model
trainer.train()
```

# testing the Summarization Logic

```python
def summarize_dialogue(dialogue: str) -> str :
    dialogue = clean_data(dialogue)  # data cleaning
    # tokenize
    inputs= tokenizer(
        dialogue,
        padding= "max_length",
        max_length= 512,
        truncation= True,
        return_tensors= 'pt'
    ).to(device)
    # generate the summary(which in the form of token_id)
    model.to(device)
    targets= model.generate(
        input_ids= inputs['input_ids'],
        attention_mask= inputs['attention_mask'],
        max_length= 150,
        num_beams= 4,
        early_stopping= True
    )
    # decoding part (summary in text format)
    summary = tokenizer.decode(targets[0], skip_special_tokens= True)
    return summary
```
